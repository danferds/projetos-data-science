```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

# Requisição e selecação

```python
# Função para realizar o WebScraping
def extrair_info_estado(estado: str) -> dict:
    print(f'[ ] Buscando informações de {estado}...')
    url_estado = f'https://www.ibge.gov.br/cidades-e-estados/{estado}.html'
    pagina = requests.get(url_estado, verify=False)

    soup = BeautifulSoup(pagina.content, 'html.parser')
    indicadores = soup.select('.indicador')

    estado_dict = {
        ind.select('.ind-label')[0].text: ind.select('.ind-value')[0].text
        for ind in indicadores
    }

    estado_dict['Estado'] = estado

    return estado_dict
```

```python
estados = ['AC', 'AL', 'AP', 'AM', 'BA', 'CE', 'DF', 'ES', 'GO', 'MA', 'MT', 'MS', 'MG', 'PA', 'PB', 'PR', 'PE', 'PI', 'RJ', 'RN', 'RS', 'RO', 'RR', 'SC', 'SP', 'SE', 'TO']

dados_estados = [extrair_info_estado(estado) for estado in estados]
```

```python
df = pd.DataFrame(dados_estados)
df.head()
```

# Limpeza

```python
estados_df = df.copy()
estados_df.columns = ['governador', 'capital', 'gentílico', 'área', 'população', 'densidade_demográfica', 'matrículas_ensino_fundamental', 'idh', 'receita_realizada', 'despesas_empenhadas', 'renda_per_capita', 'total_veículos', 'código']
estados_df = estados_df[['código', 'governador', 'população', 'área', 'idh', 'renda_per_capita', 'total_veículos', 'matrículas_ensino_fundamental', 'despesas_empenhadas', 'receita_realizada']]

estados_df = estados_df.replace({
    '\.': '',
    ',': '.',
    '\[\d+\]': '',
    ' hab/km²': '',
    ' km²': '',
    ' pessoas': '',
    ' matrículas': '',
    'R\$.*': '',
    ' veículos': ''
}, regex=True)

estados_df.head()

colunas_numericas = ['população', 'área', 'idh', 'renda_per_capita', 'total_veículos', 'matrículas_ensino_fundamental', 'despesas_empenhadas', 'receita_realizada']

estados_df[colunas_numericas] = estados_df[colunas_numericas].apply(lambda x: x.str.strip())
estados_df[colunas_numericas] = estados_df[colunas_numericas].apply(pd.to_numeric)
estados_df.info()
```
